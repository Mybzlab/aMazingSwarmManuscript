```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# tall vertical figure
fig,ax = plt.subplots(figsize=(4, 8))

Ns = [1, 2, 3, 7, 20, 50]

# measured l0
l0 = 45.42

# measured lp of the payload in different trials
N20 = np.array([121.82, 57.23, 90.35])
N50 = np.array([127.42, 110.85, 33.03])

N20_avg = np.mean(N20)
N20_SE = np.std(N20 / l0) / np.sqrt(len(N20))
N50_avg = np.mean(N50)
N50_SE = np.std(N50 / l0) / np.sqrt(len(N50))

lps = np.array([0.09, 5.65, 11.35, 15.34, N20_avg, N50_avg]) / l0
lps_yerror = [0, 0, 0, 0, N20_SE, N50_SE]


ax.errorbar(
    Ns,
    lps,
    yerr=lps_yerror,
    fmt='o-',
    capsize=5,
    linewidth=2,
    markersize=10
)
ax.set_xticks([0,25,50])
ax.tick_params(axis='both', labelsize=24)
ax.grid(True)
ax.set_xlabel(r"$N$",fontsize=24)
ax.set_ylabel(r"$\ell_p / \ell_0$",fontsize=24)
ax.set_title("Experimental payload persistence\nas a function of swarm size", pad=15,fontsize=24)
plt.tight_layout()
plt.savefig("payl_pers_vert.jpg", bbox_inches="tight", dpi=300)
plt.show()
```
